```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("hotel_bookings.csv")

missing_count = df.isnull().sum()
missing_percentage = (missing_count / len(df)) * 100

missing_table = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
}).sort_values(by="Missing Percentage", ascending=False)

display(missing_table)

df = df.drop(columns=["company"])

df = df.drop(columns=["reservation_status", "reservation_status_date"])

numerical_cols = df.select_dtypes(include=["int64", "float64"]).columns.tolist()

n_cols = 3
n_rows = (len(numerical_cols) + n_cols - 1) // n_cols

fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5 * n_rows))
axes = axes.flatten()

for i, col in enumerate(numerical_cols):
    axes[i].boxplot(df[col].dropna())
    axes[i].set_title(col)
    axes[i].set_ylabel(col)

for i in range(len(numerical_cols), len(axes)):
    axes[i].axis("off")

plt.tight_layout()
plt.show()

outlier_results = []

for col in numerical_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1

    lower_fence = q1 - 1.5 * iqr
    upper_fence = q3 + 1.5 * iqr

    outlier_count = (
        (df[col] < lower_fence) |
        (df[col] > upper_fence)
    ).sum()

    outlier_results.append({
        "Column": col,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Lower Fence": lower_fence,
        "Upper Fence": upper_fence,
        "IQR Outlier Count": outlier_count,
        "IQR Outlier Percentage": (outlier_count / len(df)) * 100
    })

outlier_table = pd.DataFrame(outlier_results)

display(outlier_table)

percentile_cols = [
    "lead_time",
    "stays_in_weekend_nights",
    "stays_in_week_nights",
    "adr"
]

percentile_mask = pd.Series(False, index=df.index)

for col in percentile_cols:
    lower = df[col].quantile(0.01)
    upper = df[col].quantile(0.99)

    percentile_mask |= (
        (df[col] < lower) |
        (df[col] > upper)
    ).fillna(False)

    print(col)
    print("1st percentile:", lower)
    print("99th percentile:", upper)
    print()

print("Rows before percentile trimming:", len(df))
print("Rows identified for removal:", percentile_mask.sum())

df = df.loc[~percentile_mask].copy()

print("Rows after percentile trimming:", len(df))
print("Total rows removed:", percentile_mask.sum())

missing_after = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df)) * 100
}).sort_values(by="Missing Percentage", ascending=False)

print("\nMissing values after cleaning:")
display(missing_after)

df.to_csv("hotel_bookings_cleaned.csv", index=False)

print("Cleaned dataset saved as hotel_bookings_cleaned.csv")

# Justification for dropping column is simply very less values, although
#  company might show a correlation, for eg, tickets booked by company are
#  less likely to be cancelled, but with limited data, any meaningful
#  predictions are difficult and can affect our model for other cases too.
#
# We kept 'Agent' column because 86%+ values are available and it can
#  be valuable for analysis.
#
# Outliers were not removed by iqr formula and rather for some continuous
#  numeric ones beyond range 1th-99th percentile because of 3 main reasons:
# 1) Data is heavily skewed with many attributes having mean '0'.
#    Removing outliers can remove valuable information too.
# 2) An outlier in 1 attribute can remove full row, where other row
#    information can prove out to be valuable. A naive analysis 
#    removed 50% of data from boxplot using 1.5*IQR+Qi method.
# 3) Skewed data itself changes standard deviation itself which makes
#    it less reliable(lof, as discussed in class could have been used).
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import KNNImputer, SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OneHotEncoder

X = df.drop(columns=["is_canceled"])
y = df["is_canceled"]

numerical_cols = X.select_dtypes(include=["int64", "float64"]).columns.tolist()
categorical_cols = X.select_dtypes(include=["object"]).columns.tolist()

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

numerical_pipeline_a = Pipeline([
    ("imputer", KNNImputer(n_neighbors=5)),
    ("scaler", StandardScaler())
])

numerical_pipeline_b = Pipeline([
    ("imputer", KNNImputer(n_neighbors=5)),
    ("scaler", MinMaxScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor_a = ColumnTransformer([
    ("numerical", numerical_pipeline_a, numerical_cols),
    ("categorical", categorical_pipeline, categorical_cols)
])

preprocessor_b = ColumnTransformer([
    ("numerical", numerical_pipeline_b, numerical_cols),
    ("categorical", categorical_pipeline, categorical_cols)
])

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nX_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)

print("\ny_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)

print("\nNumerical columns:")
print(numerical_cols)

print("\nCategorical columns:")
print(categorical_cols)

print("\nPipeline A:")
print(preprocessor_a)

print("\nPipeline B:")
print(preprocessor_b)

print("\nTask 3 completed successfully.")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.pipeline import Pipeline
import time

# Create the four pipelines

logistic_standard = Pipeline([
    ("preprocessor", preprocessor_a),
    ("model", LogisticRegression(max_iter=1000))
])

logistic_minmax = Pipeline([
    ("preprocessor", preprocessor_b),
    ("model", LogisticRegression(max_iter=1000))
])

tree_standard = Pipeline([
    ("preprocessor", preprocessor_a),
    ("model", DecisionTreeClassifier(random_state=42))
])

tree_minmax = Pipeline([
    ("preprocessor", preprocessor_b),
    ("model", DecisionTreeClassifier(random_state=42))
])

print("All four pipelines created successfully.")


# Train Logistic Regression

print("\n1/4 Training Logistic Regression + StandardScaler...")
start = time.time()

logistic_standard.fit(X_train, y_train)

print(f"Completed in {time.time() - start:.2f} seconds")


print("\n2/4 Training Logistic Regression + MinMaxScaler...")
start = time.time()

logistic_minmax.fit(X_train, y_train)

print(f"Completed in {time.time() - start:.2f} seconds")


# Train Decision Tree

print("\n3/4 Training Decision Tree + StandardScaler...")
start = time.time()

tree_standard.fit(X_train, y_train)

print(f"Completed in {time.time() - start:.2f} seconds")


print("\n4/4 Training Decision Tree + MinMaxScaler...")
start = time.time()

tree_minmax.fit(X_train, y_train)

print(f"Completed in {time.time() - start:.2f} seconds")


print("\nAll four models have been trained successfully.")
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

models = {
    "Logistic Regression + StandardScaler": logistic_standard,
    "Logistic Regression + MinMaxScaler": logistic_minmax,
    "Decision Tree + StandardScaler": tree_standard,
    "Decision Tree + MinMaxScaler": tree_minmax
}

results = []

for name, model in models.items():
    # Predictions
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)

    # Metrics
    train_accuracy = accuracy_score(y_train, y_train_pred)
    test_accuracy = accuracy_score(y_test, y_test_pred)
    precision = precision_score(y_test, y_test_pred, zero_division=0)
    recall = recall_score(y_test, y_test_pred, zero_division=0)
    f1 = f1_score(y_test, y_test_pred, zero_division=0)

    results.append({
        "Model-Pipeline": name,
        "Training Accuracy": train_accuracy,
        "Testing Accuracy": test_accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1,
        "Train-Test Gap": train_accuracy - test_accuracy
    })

comparison_table = pd.DataFrame(results)

# Round values for presentation
comparison_table_display = comparison_table.copy()

metric_columns = [
    "Training Accuracy",
    "Testing Accuracy",
    "Precision",
    "Recall",
    "F1-Score",
    "Train-Test Gap"
]

comparison_table_display[metric_columns] = (
    comparison_table_display[metric_columns].round(4)
)

display(comparison_table_display)
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# Logistic Regression

best_logistic = logistic_standard

y_pred_logistic = best_logistic.predict(X_test)

cm_logistic = confusion_matrix(y_test, y_pred_logistic)

print("Confusion Matrix - Best Logistic Regression")
print(cm_logistic)

fig, ax = plt.subplots(figsize=(6, 5))

ConfusionMatrixDisplay(
    confusion_matrix=cm_logistic,
    display_labels=["Not Canceled (0)", "Canceled (1)"]
).plot(
    ax=ax,
    values_format="d",
    colorbar=False
)

ax.set_title("Best Logistic Regression - Confusion Matrix")
plt.tight_layout()

plt.savefig(
    "confusion_matrix_logistic_regression.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# Decision tree

best_tree = tree_standard

y_pred_tree = best_tree.predict(X_test)

cm_tree = confusion_matrix(y_test, y_pred_tree)

print("Confusion Matrix - Best Decision Tree")
print(cm_tree)

fig, ax = plt.subplots(figsize=(6, 5))

ConfusionMatrixDisplay(
    confusion_matrix=cm_tree,
    display_labels=["Not Canceled (0)", "Canceled (1)"]
).plot(
    ax=ax,
    values_format="d",
    colorbar=False
)

ax.set_title("Best Decision Tree - Confusion Matrix")
plt.tight_layout()

plt.savefig(
    "confusion_matrix_decision_tree.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# The Decision Tree shows possible overfitting because its training accuracy is 0.9960,
# while its testing accuracy is only about 0.8591, giving a large train-test gap of 0.1369.
# This means it fits training data extremely well but performs noticeably worse on unseen data.

# In contrast, Logistic Regression shows little evidence of overfitting, with train-test
# gaps of only 0.0064 (StandardScaler) and 0.0058 (MinMaxScaler).
```

## Task 6 - Final Observations

1. **Best overall result:** Decision Tree with StandardScaler gives the best overall performance, with a testing accuracy of **85.91%**, precision of **80.96%**, recall of **80.38%**, and F1-score of **80.67%**.

2. **Effect of scaling on Logistic Regression:** StandardScaler performs slightly better than MinMaxScaler for Logistic Regression. The F1-score is **71.71%** with StandardScaler compared with **70.99%** with MinMaxScaler. Therefore, scaling has a noticeable but relatively small effect on Logistic Regression.

3. **Effect of scaling on Decision Tree:** Scaling makes almost no difference to the Decision Tree. The F1-score is **80.67%** with StandardScaler and **80.64%** with MinMaxScaler. This is expected because Decision Trees are generally not sensitive to feature scale.

4. **Overfitting:** The Decision Tree shows possible overfitting because its training accuracy is **99.60%**, while its testing accuracy is **85.91%**, resulting in a train-test accuracy gap of **13.69 percentage points**. In comparison, Logistic Regression has much smaller gaps of **0.64** and **0.58 percentage points**, indicating better generalization.

5. **Recall and confusion matrices:** The Decision Tree has a higher recall (**80.38%**) than Logistic Regression (**65.24%**), meaning it identifies more of the actual canceled bookings. The Decision Tree has **1,674 false negatives**, compared with **2,966** for Logistic Regression. Therefore, the Decision Tree is better at detecting canceled bookings, although its larger train-test gap indicates possible overfitting.

### Extension

A useful extension would be to investigate whether extreme outliers contribute to Decision Tree overfitting. The Decision Tree could be evaluated before and after different outlier-treatment strategies while keeping the same train-test split and model settings. The changes in testing F1-score, recall, and the train-test accuracy gap could then be compared. If outlier removal reduces the accuracy gap while maintaining or improving recall, it would suggest that extreme observations contributed to the overfitting. However, the current results alone do not prove that outliers are the cause of the overfitting.
