```python
# Task 2

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("\nA:")
print(A)

print("\nB:")
print(B)

addition = A + B
print("\nAddition:")
print(addition)

element_multiplication = A * B
print("\nElement-wise multiplication:")
print(element_multiplication)

matrix_multiplication = A @ B
print("\nMatrix multiplication:")
print(matrix_multiplication)

transpose_A = A.T
print("\nTranspose of A:")
print(transpose_A)

determinant_A = np.linalg.det(A)
print("\nDeterminant of A:")
print(determinant_A)

if determinant_A != 0:
    inverse_A = np.linalg.inv(A)

    print("\nInverse of A:")
    print(inverse_A)

    result = A @ inverse_A

    print("\nA @ A^-1:")
    print(result)

    print("\nCheck:")
    print(np.allclose(result, np.eye(A.shape[0])))

else:
    print("\nA is not invertible.")
```

```python
# Task 4

import pandas as pd

df = pd.read_csv("train.csv")

print("\nFirst 5 rows:")
print(df.head())

print("\nLast 5 rows:")
print(df.tail())

print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nInfo:")
df.info()

print("\nDescription:")
print(df.describe())


# loc example
print("\nUsing loc:")
print(df.loc[0:4, ["Name", "Sex", "Age"]])


# iloc example
print("\nUsing iloc:")
print(df.iloc[0:5, 3:6])


print("""
Difference between loc and iloc:

loc is used for label-based selection while iloc is used for position-based selection.
""")
```

```python
# Task 5

# Male passengers older than 50
male_over_50 = df[(df["Sex"] == "male") & (df["Age"] > 50)]

print("\nMale passengers older than 50:")
print(len(male_over_50))


# Female first-class passengers
female_first_class = df[(df["Sex"] == "female") & (df["Pclass"] == 1)]

female_count = len(female_first_class)
female_survival_rate = female_first_class["Survived"].mean() * 100

print("\nFemale first-class passengers:")
print("Count:", female_count)
print("Survival rate:", female_survival_rate, "%")


# Age 20-40, Fare above overall median, and survived
fare_median = df["Fare"].median()

age_fare_survived = df[
    (df["Age"] >= 20) &
    (df["Age"] <= 40) &
    (df["Fare"] > fare_median) &
    (df["Survived"] == 1)
]

print("\nAge 20-40, fare above median, and survived:")
print("Count:", len(age_fare_survived))


# Travelling alone, age below 30, and did not survive
alone_under_30 = df[
    (df["SibSp"] == 0) &
    (df["Parch"] == 0) &
    (df["Age"] < 30) &
    (df["Survived"] == 0)
]

print("\nTravelling alone, age below 30, and did not survive:")
print("Count:", len(alone_under_30))


# Embarked S, Pclass 2 or 3, and Fare above Southampton median
southampton = df[df["Embarked"] == "S"]
southampton_median = southampton["Fare"].median()

southampton_filter = df[
    (df["Embarked"] == "S") &
    (df["Pclass"].isin([2, 3])) &
    (df["Fare"] > southampton_median)
]

print("\nEmbarked S, Pclass 2 or 3, and fare above Southampton median:")
print("Count:", len(southampton_filter))
```

```python
# Task 6

# Survival rate by Sex
survival_by_sex = df.groupby("Sex")["Survived"].mean()

print("\nSurvival rate by Sex:")
print(survival_by_sex)


# Survival rate by Pclass
survival_by_pclass = df.groupby("Pclass")["Survived"].mean()

print("\nSurvival rate by Pclass:")
print(survival_by_pclass)


# Average Age and Fare by Pclass
age_fare_by_pclass = df.groupby("Pclass")[["Age", "Fare"]].mean()

print("\nAverage Age and Fare by Pclass:")
print(age_fare_by_pclass)


# Passenger count and survival rate by Sex-Pclass
sex_pclass = df.groupby(["Sex", "Pclass"]).agg(
    Passenger_Count=("PassengerId", "count"),
    Survival_Rate=("Survived", "mean")
)

print("\nPassenger count and survival rate by Sex-Pclass:")
print(sex_pclass)


# Passenger count, average Fare, and survival rate by Embarked
embarked_stats = df.groupby("Embarked").agg(
    Passenger_Count=("PassengerId", "count"),
    Average_Fare=("Fare", "mean"),
    Survival_Rate=("Survived", "mean")
)

print("\nPassenger count, average Fare, and survival rate by Embarked:")
print(embarked_stats)
```

```python
# Task 7

import matplotlib.pyplot as plt


# Missing values
missing_count = df.isnull().sum()
missing_percentage = (missing_count / len(df)) * 100

print("\nMissing count:")
print(missing_count)

print("\nMissing percentage:")
print(missing_percentage)


# Plot missing values
plt.figure(figsize=(10, 5))
plt.bar(missing_count.index, missing_count.values)

plt.title("Missing Values")
plt.xlabel("Columns")
plt.ylabel("Count")
plt.xticks(rotation=45)

plt.show()


# Fill missing Age with mean
print("\nMissing Age before filling:")
print(df["Age"].isnull().sum())

age_mean = df["Age"].mean()

df["Age"] = df["Age"].fillna(age_mean)

print("Missing Age after filling:")
print(df["Age"].isnull().sum())


# Age imputation methods
age = pd.read_csv("train.csv")["Age"]

age_mean_data = age.fillna(age.mean())
age_median_data = age.fillna(age.median())
age_mode_data = age.fillna(age.mode()[0])

np.random.seed(42)

age_random_data = age.copy()
age_values = age.dropna()

missing = age_random_data.isnull()

age_random_data.loc[missing] = np.random.choice(
    age_values,
    size=missing.sum()
)

print("\nMissing Age after different methods:")
print("Mean:", age_mean_data.isnull().sum())
print("Median:", age_median_data.isnull().sum())
print("Mode:", age_mode_data.isnull().sum())
print("Random:", age_random_data.isnull().sum())


# Fare outliers
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["Fare"] < lower_bound) |
    (df["Fare"] > upper_bound)
]

print("\nFare outliers:")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Number of outliers:", len(outliers))
```

```python
# Task 8

# Create FamilySize
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

# Create IsAlone
df["IsAlone"] = (df["FamilySize"] == 1).astype(int)

print("\nFamilySize and IsAlone:")
print(df[["SibSp", "Parch", "FamilySize", "IsAlone"]].head())


# Create pivot table
pivot = pd.pivot_table(
    df,
    values="Survived",
    index="Sex",
    columns="Pclass",
    aggfunc="mean"
)

print("\nSurvival rate by Sex and Pclass:")
print(pivot)


# Find highest and lowest groups
highest_group = pivot.stack().idxmax()
lowest_group = pivot.stack().idxmin()

highest_rate = pivot.stack().max()
lowest_rate = pivot.stack().min()

print("\nHighest survival group:")
print(highest_group, ":", highest_rate)

print("\nLowest survival group:")
print(lowest_group, ":", lowest_rate)
```

```python
# Task 9

import matplotlib.pyplot as plt
import seaborn as sns


# --------------------------------------------------
# 1. Correlation heatmap
# --------------------------------------------------

numeric_columns = [
    "Survived",
    "Pclass",
    "Age",
    "SibSp",
    "Parch",
    "Fare",
    "FamilySize",
    "IsAlone"
]

correlation = df[numeric_columns].corr()

print("\nCorrelation:")
print(correlation.round(2))

plt.figure(figsize=(10, 7))
sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap")
plt.show()


# Find strongest positive and negative relationships

corr_values = correlation.copy()

for column in corr_values.columns:
    corr_values.loc[column, column] = np.nan

positive_pair = corr_values.unstack().dropna().idxmax()
positive_value = corr_values.unstack().dropna().max()

negative_pair = corr_values.unstack().dropna().idxmin()
negative_value = corr_values.unstack().dropna().min()

print("\nStrongest positive relationship:")
print(positive_pair, "=", round(positive_value, 3))

print("\nStrongest negative relationship:")
print(negative_pair, "=", round(negative_value, 3))


# --------------------------------------------------
# 2. Survival rate by Sex
# --------------------------------------------------

survival_by_sex = df.groupby("Sex")["Survived"].mean() * 100

print("\nSurvival rate by Sex:")
print(survival_by_sex.round(2))

plt.figure(figsize=(7, 5))
plt.bar(
    survival_by_sex.index,
    survival_by_sex.values
)

plt.title("Survival Rate by Sex")
plt.xlabel("Sex")
plt.ylabel("Survival Rate (%)")

plt.show()


# --------------------------------------------------
# 3. Age vs Fare
# --------------------------------------------------

survived = df[df["Survived"] == 1]
not_survived = df[df["Survived"] == 0]

plt.figure(figsize=(9, 6))

plt.scatter(
    not_survived["Age"],
    not_survived["Fare"],
    alpha=0.5,
    label="Did not survive"
)

plt.scatter(
    survived["Age"],
    survived["Fare"],
    alpha=0.5,
    label="Survived"
)

plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.legend()

plt.show()


# --------------------------------------------------
# 4. Some additional values for observations
# --------------------------------------------------

overall_survival = df["Survived"].mean() * 100

female_survival = survival_by_sex["female"]
male_survival = survival_by_sex["male"]

pclass_survival = df.groupby("Pclass")["Survived"].mean() * 100

average_fare = df.groupby("Pclass")["Fare"].mean()

alone_survival = df.groupby("IsAlone")["Survived"].mean() * 100


# --------------------------------------------------
# 5. Observations
# --------------------------------------------------

print("\nObservations:")
print(
    f"1. The overall survival rate was approximately "
    f"{overall_survival:.2f}%."
)

print(
    f"2. Female passengers had a survival rate of "
    f"{female_survival:.2f}%, compared with "
    f"{male_survival:.2f}% for male passengers."
)

print(
    f"3. The highest survival rate by passenger class was "
    f"{pclass_survival.idxmax()} class at "
    f"{pclass_survival.max():.2f}%."
)

print(
    f"4. The lowest survival rate by passenger class was "
    f"{pclass_survival.idxmin()} class at "
    f"{pclass_survival.min():.2f}%."
)

print(
    f"5. The strongest positive correlation among the selected "
    f"numerical variables was between {positive_pair[0]} and "
    f"{positive_pair[1]} ({positive_value:.2f})."
)

print(
    f"6. The strongest negative correlation was between "
    f"{negative_pair[0]} and {negative_pair[1]} "
    f"({negative_value:.2f})."
)

print(
    f"7. Passengers travelling alone had a survival rate of "
    f"{alone_survival[1]:.2f}%, while passengers not travelling "
    f"alone had a survival rate of {alone_survival[0]:.2f}%."
)
```
