```python
import numpy as np
import pandas as pd
import math


# 1. Calculate entropy
def entropy(df):
    counts = df["class"].value_counts().to_numpy()
    probabilities = counts / len(df)

    return -sum(
        p * math.log2(p)
        for p in probabilities
        if p > 0
    )


# 2. Calculate Gini impurity
def gini(df):
    probabilities = df["class"].value_counts().to_numpy() / len(df)
    return 1 - np.sum(probabilities ** 2)


# 3. Calculate impurity according to the selected criterion
def impurity(df, criterion="gini"):
    if len(df) == 0:
        return 0

    if criterion == "gini":
        return gini(df)
    elif criterion == "entropy":
        return entropy(df)
    else:
        raise ValueError("criterion must be 'gini' or 'entropy'")


# 4. Find the best feature and threshold
def split(df, criterion="gini"):
    parent_impurity = impurity(df, criterion)

    best_gain = 0
    best_feature = None
    best_threshold = None
    best_left = None
    best_right = None

    if len(df) < 2 or parent_impurity == 0:
        return best_gain, best_feature, best_threshold, best_left, best_right

    features = [c for c in df.columns if c != "class"]

    for feature in features:
        values = np.sort(df[feature].unique())

        # Candidate thresholds: midpoints between unique values
        thresholds = (values[:-1] + values[1:]) / 2

        for threshold in thresholds:
            left = df[df[feature] <= threshold]
            right = df[df[feature] > threshold]

            if len(left) == 0 or len(right) == 0:
                continue

            weighted_impurity = (
                len(left) / len(df) * impurity(left, criterion)
                + len(right) / len(df) * impurity(right, criterion)
            )

            gain = parent_impurity - weighted_impurity

            if gain > best_gain:
                best_gain = gain
                best_feature = feature
                best_threshold = threshold
                best_left = left
                best_right = right

    return (
        best_gain,
        best_feature,
        best_threshold,
        best_left,
        best_right
    )


# 5. Build the tree recursively
def build_tree(
    df,
    criterion="gini",
    max_depth=5,
    min_samples_split=2,
    depth=0
):
    # Predict the majority class at this node
    majority_class = int(df["class"].mode().iloc[0])

    # Stopping condition 1: pure node
    if df["class"].nunique() == 1:
        return {"leaf": True, "prediction": majority_class}

    # Stopping condition 2: maximum depth reached
    if depth >= max_depth:
        return {"leaf": True, "prediction": majority_class}

    # Stopping condition 3: too few samples to split
    if len(df) < min_samples_split:
        return {"leaf": True, "prediction": majority_class}

    gain, feature, threshold, left, right = split(df, criterion)

    # Stopping condition 4: no valid improvement
    if feature is None or gain <= 1e-12:
        return {"leaf": True, "prediction": majority_class}

    # Build both subtrees
    left_child = build_tree(
        left, criterion, max_depth, min_samples_split, depth + 1
    )
    right_child = build_tree(
        right, criterion, max_depth, min_samples_split, depth + 1
    )

    return {
        "leaf": False,
        "feature": feature,
        "threshold": float(threshold),
        "gain": float(gain),
        "left": left_child,
        "right": right_child
    }


# 6. Predict one sample
def predict_one(tree, row):
    if tree["leaf"]:
        return tree["prediction"]

    if row[tree["feature"]] <= tree["threshold"]:
        return predict_one(tree["left"], row)

    return predict_one(tree["right"], row)


# 7. Predict multiple samples
def predict_tree(tree, X):
    return np.array([
        predict_one(tree, row)
        for row in X.to_dict(orient="records")
    ])


# 8. One function to train and predict
def decision_tree_classifier(
    train_df,
    test_df,
    criterion="gini",
    max_depth=5,
    min_samples_split=2
):
    tree = build_tree(
        train_df,
        criterion=criterion,
        max_depth=max_depth,
        min_samples_split=min_samples_split
    )

    predictions = predict_tree(tree, test_df.drop(columns=["class"]))

    return tree, predictions
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

train_df, test_df = train_test_split(
    df,
    test_size=0.2,
    random_state=42,
    stratify=df["class"]
)

tree, predictions = decision_tree_classifier(
    train_df,
    test_df,
    criterion="gini",
    max_depth=5
)

print("Manual tree accuracy:", accuracy_score(
    test_df["class"], predictions
))
```
